```python
import numpy as np

# Sample data for demonstration. You can replace this with your actual data.
data = np.array([10, 25, 5, 30, 15, 20])

# Values greater than 20
print("Values greater than 20:", data[data > 20])

# Count values less than 20
print("Count less than 20:", np.sum(data < 20))

# Replace values less than 20 with 20
modified_data = data.copy()
modified_data[modified_data < 20] = 20
print("Modified data:", modified_data)
```
